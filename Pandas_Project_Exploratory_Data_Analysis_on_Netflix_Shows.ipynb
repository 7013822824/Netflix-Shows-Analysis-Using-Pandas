```python
import pandas as pd
df = pd.read_csv('/content/sample_data/netflix_titles.csv')
```

```python
df.info()
df.describe()
df.head()
```

```python
df.dropna(subset=["title", "type", "release_year"], inplace=True)
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), format='%B %d, %Y', errors='coerce')
```

```python
df['type'].value_counts()
```

```python
df['country'].value_counts().head(10)
```

```python
df['year_added'] = df['date_added'].dt.year
df['year_added'].value_counts().sort_index().plot(kind='bar', figsize=(10,5))
```

```python
df['director'].value_counts().head(10)
```

```python
df['listed_in'].str.split(', ').explode().value_counts().head(10)
```
